## Переименование столбцов

```python
main_columns = {
    "SKU": "sku",
    "Категория": "category",
    "Предмет": "product_type",
    "Бренд": "brand",
    "Страна": "country"
}

metrics = {
    "Цена": "price",
    "Продажи": "sales_sum",
    "Выручка": "revenue_sum",
    "Средний % выкупа": "avg_buyout"
}

months = {
    "Ноябрь": "november",
    "Декабрь": "december"
}


def rename_column(column):
    if column in main_columns:
        return main_columns[column]

    if "(" in column and ")" in column:
        metric, month = column.split("(")
        metric = metrics.get(metric.strip())
        month = months.get(month.strip(") "))

        if metric and month:
            return f"{metric}_{month}"

    return column
```

```python
sales = sales.rename(columns=rename_column)
sales.head()
```

```python
brands = brands.rename(columns={
    "Бренд": "brand",
    "Месяц": "month",
    "Продажи": "sales_sum",
    "Выручка": "revenue_sum",
    "Средний % выкупа": "avg_buyout"
})

brands.head()
```

## Преобразование таблицы Product

Перевожу показатели ноября и декабря из отдельных столбцов в длинный формат.

```python
id_columns = [
    "sku",
    "category",
    "product_type",
    "brand",
    "country"
]

sales_long = sales.melt(
    id_vars=id_columns,
    var_name="metric_month",
    value_name="value"
)

sales_long[["metric", "month_name"]] = (
    sales_long["metric_month"]
    .str.rsplit("_", n=1, expand=True)
)

sales_long = (
    sales_long
    .pivot_table(
        index=id_columns + ["month_name"],
        columns="metric",
        values="value",
        aggfunc="first"
    )
    .reset_index()
)

sales_long.columns.name = None

sales_long = sales_long.rename(columns={
    "sales_sum": "sales",
    "revenue_sum": "revenue"
})

month_numbers = {
    "november": 11,
    "december": 12
}

sales_long["month"] = pd.to_datetime({
    "year": 2023,
    "month": sales_long["month_name"].map(month_numbers),
    "day": 1
})

sales = sales_long.drop(columns="month_name")
```
